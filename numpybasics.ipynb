```python
arr1 = np.array([1.123456, 2.789101, 0.0000001])
print(np.array2string(arr1))
```

```python
arr1 = np.array([13, 14, 3])
arr2 = np.array([17, 13, 13])

result = np.bitwise_and(arr1, arr2)
print(result)
```

```python
pos_inf = np.inf
neg_inf = -np.inf
arr = np.array([1.0, np.inf, -np.inf, 4.5])
print(np.isinf(arr))    
print(np.isposinf(arr))    
print(np.isfinite(arr))
```
